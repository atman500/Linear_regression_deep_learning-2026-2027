```python
import numpy as  np
import pandas as pd
import  matplotlib.pyplot as plt
```

```python
x=[1,2,3,4,5,6,7,8]
y=[12,11,13,10,16,17,14,18]
```

```python
df=pd.DataFrame({
    "x": x,
    "y" :y
})
```

```python
print(df.head())
```

```python
import statsmodels.api as sm
```

```python
import numpy as np
import statsmodels.api as sm

# إضافة الثابت للمتغير المستقل
X_const = sm.add_constant(x)
linear_model = sm.OLS(y, X_const).fit()

print(linear_model.summary())
```

```python
y_pred=linear_model.predict(X_const)
```

```python

# (Y_pred)
y_pred = linear_model.predict(X_const)
# MSE و RMSE
mse = np.mean((np.array(y) - y_pred) ** 2)
rmse = np.sqrt(mse)

print(f"Mean Squared Error (MSE): {mse}")
print(f"Root Mean Squared Error (RMSE): {rmse}")
```
