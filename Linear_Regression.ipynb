```python
x=[1,2,3,4,5,6,7,8]
y=[12,11,13,10,16,17,14,18]
print(x,y)
```

developpe a linear regression **model** **bold text**

```python
import  statsmodels.api as sm
```

```python
model=sm.OLS(y,x).fit()
x=sm.add_constant(x)
print(model.summary())
```

```python
linear_model=sm.OLS(y,x).fit()
print(linear_model.summary())
```

```python
y_pred=linear_model.predict(x)
```

```python
MSE=np.mean((y-y_pred))/len(y)
```

```python
RMSE=np.square(MSE)
```

```python
print("the avlues  of  MSE is  equale to :",MSE)
print("the value of  RMSE is  equale to :",RMSE)
```
