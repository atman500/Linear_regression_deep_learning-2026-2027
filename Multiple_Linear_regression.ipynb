```python
df=pd.read_csv('mult.csv',sep=',')
print(df.head())
```

```python
x=df[['Study_Hours','Attendance_Percent','Assignments_Completed','Previous_Score','Sleep_Hours']]
y=df['Exam_Score']
print(x.head())
print(y.head())
```

```python
import statsmodels.api as sm
```

```python
mult_model=sm.OLS(y,x).fit()
x=sm.add_constant(x)
print(mult_model.summary())
```

```python
y_pred=mult_model.predict(x)
print(y_pred.head())
print(y.head())
DF=pd.DataFrame({
    "actual_y":y,
    "predict_y":y_pred,
    "error":y-y_pred
})
print(DF.head())
```

```python
MSE=mult_model.mse_resid
print('MSE_value is  equla to :',MSE)

RMSE=np.sqrt(MSE)
print('RMSE_value is equale to:',RMSE)
```
