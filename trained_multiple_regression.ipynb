```python
df=pd.read_csv('Multiplt_Linear_Regression.csv',sep=',')
print(df.head())
```

```python
x=df.drop(columns=['Exam_Score'],axis=1)
y=df['Exam_Score']
```

```python

print(y.head())
print(x.head())
```

```python
import sklearn
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
from sklearn.linear_model import LinearRegression
model=LinearRegression()
trained_model=model.fit(x_train,y_train)
```

```python
print(trained_model.coef_)
print(trained_model.intercept_)
```

```python
import statsmodels.api as sm
intercept = sm.add_constant(x_train)
# Create and fit the OLS model
model_sm = sm.OLS(y_train, intercept)
results = model_sm.fit()
# Print the summary
print(results.summary())
```

```python
y_pred = trained_model.predict(x_test)
print(y_pred[0:5])
```
